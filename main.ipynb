```python
from ultralytics import YOLO
import cv2

model_path = 'last.pt'
image_path = 'MAH03602.MP4'

model=YOLO(model_path)
model.predict(source=image_path,
              show=True,
              conf=0.25,
              boxes=True,
              #retina_masks=True,
              vid_stride=True,
              classes=[0,2,3],
              visualize=True
              )
```
